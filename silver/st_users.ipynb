```python
class StUsers(LakehouseMaster):
    
    def transform(self, df):
        # Deduplicación
        df_clean = df.dropDuplicates(["id"])
        
        # Limpieza genérica
        df_clean = AnalyticsUtils.cleaning_df_spark(df_clean)
        
        # Gobernanza PII
        df_clean = AnalyticsGobernanzaHelpers.enmascarar_pii_sha256(df_clean, ["email"])
        
        return df_clean
```
