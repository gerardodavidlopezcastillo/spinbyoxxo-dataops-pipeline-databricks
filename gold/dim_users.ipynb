```python
from src.etl_master import LakehouseMaster
from pyspark.sql.functions import col
```

```python
class DimUsers(LakehouseMaster):
    def transform(self, df):
        self.logger.info("Construyendo dimensión de usuarios")
        return df.select(
            col("id").alias("user_sk"),
            "name", 
            "country", 
            "email_hash", 
            "created_at"
        )
```
